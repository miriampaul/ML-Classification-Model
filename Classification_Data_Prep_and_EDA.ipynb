```python
dm_df = pd.read_csv('/content/dm (1).csv')
sv_df = pd.read_csv('/content/sv (2).csv')

print('DM DataFrame head:')
display(dm_df.head())

print('\nSV DataFrame head:')
display(sv_df.head())
```

```python
# Merge AE with DM
sv_dm_merged = pd.merge(dm_df, sv_df, on='USUBJID', how='left')

# Merge the result with SV
final_merged_df = pd.merge(sv_dm_merged, AE, on='USUBJID', how='left')

print('Head of the final merged DataFrame:')
display(final_merged_df.head())
```
